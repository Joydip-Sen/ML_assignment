```python
import numpy as np
import pandas as pd
from collections import Counter
```

```python
def d(x,y):
  z=(x-y)
  z=np.square(z)
  return np.sqrt(np.sum(z,axis=0))
```

```python
def eucledian_distance(point_a,point_b):
      return np.linalg.norm(point_b-point_a,axis=1)
```

```python
def predict_single(x_query,X_train,y_train,k):
  distances = eucledian_distance(x_query,X_train)
  k_indices = np.argsort(distances)[:k]
  k_nearest_labels = y_train[k_indices]
  most_common = Counter(k_nearest_labels).most_common(1)
  return most_common[0][0]
```

```python
def knn_predict(X_train,y_train,X_test,k=3):
  return np.array([predict_single(x,X_train,y_train,k) for x in X_test])
```
